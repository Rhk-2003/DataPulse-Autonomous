```python
from pathlib import Path

file_path = Path(
    r"C:\Users\himni\Downloads\ilovepdf_pages-to-jpg (1)\BDA Data\TLC_Data\green_tripdata_2016-01.parquet"
)

print("File exists:", file_path.exists())

if file_path.exists():
    print("Filename:", file_path.name)
    print("Size (MB):", round(file_path.stat().st_size / (1024 ** 2), 2))
    print("Extension:", file_path.suffix)
else:
    print("Check the path and confirm Jupyter is running on your Windows computer.")
```

```python
import pyarrow.parquet as pq

parquet_file = pq.ParquetFile(file_path)
metadata = parquet_file.metadata

print("Number of rows:", metadata.num_rows)
print("Number of columns:", metadata.num_columns)
print("Number of row groups:", metadata.num_row_groups)
print("\nSchema:")
print(parquet_file.schema_arrow)
```

```python

import pandas as pd

sample = pd.read_parquet(file_path, engine="pyarrow").head(10)

display(sample)
print("\nSample shape:", sample.shape)
print("\nData types:")
print(sample.dtypes)
```

```python

table = pq.read_table(file_path)
df = table.to_pandas()

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nMissing values by column:")
missing = df.isna().sum().sort_values(ascending=False)
display(missing[missing > 0])

print("\nDuplicate rows:", df.duplicated().sum())

print("\nNumeric summary:")
display(df.describe().T)
```
